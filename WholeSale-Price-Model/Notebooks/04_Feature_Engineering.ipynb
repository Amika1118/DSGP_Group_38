```python
import pandas as pd
import numpy as np
import os
```

```python
input_path = "../data/preprocessed/vegetable_prices_clean.csv"
df = pd.read_csv(input_path)

print(f"Original shape: {df.shape}")
df.head()
```

```python
columns_to_drop = ['week', 'year_week',]
df.drop(columns=columns_to_drop, inplace=True, errors='ignore')

print("Remaining columns:", list(df.columns))
df.head()
```

```python
class_mapping = {'normal': 0, 'flood_risk': 1, 'drought': -1}
cities = ['Badulla', 'Hambantota', 'Jaffna', 'Kurunegala', 'Matale', 'Nuwara_Eliya', 'Ratnapura']
actual_cols = [f"{city}_actual_class" for city in cities]

for col in actual_cols:
    if col in df.columns:
        df[col] = df[col].map(class_mapping)

print("Unique values after encoding:")
for col in actual_cols:
    if col in df.columns:
        print(f"{col}: {df[col].unique()[:5]}")
df.head()
```

```python
veg_mapping = {
    'Bitter Gourd': 1,
    'Brinjals': 2,
    'Cabbage': 3,
    'Carrot': 4,
    'Pumpkin': 5,
    'Tomatoes': 6
}
if 'vegetable' in df.columns:
    df['vegetable'] = df['vegetable'].map(veg_mapping)
    print("Unique vegetable codes:", df['vegetable'].unique())
else:
    print("Warning: 'vegetable' column not found.")
df.head()
```

```python
# For each vegetable, shift price to get previous weeks' prices
df['price_lag1'] = df.groupby('vegetable')['price'].shift(1)
df['price_lag2'] = df.groupby('vegetable')['price'].shift(2)

print("Lag features created.")
df[['vegetable', 'week_num', 'price', 'price_lag1', 'price_lag2']].head(10)
# Rolling mean and standard deviation over 4 weeks (minimum 1 observation)
df['price_roll_mean_4'] = df.groupby('vegetable')['price'].transform(
    lambda x: x.rolling(4, min_periods=1).mean()
)
df['price_roll_std_4'] = df.groupby('vegetable')['price'].transform(
    lambda x: x.rolling(4, min_periods=1).std()
)

print("Rolling features created.")
df[['vegetable', 'week_num', 'price', 'price_roll_mean_4', 'price_roll_std_4']].head(10)
```

```python
prob_types = ['prob_drought', 'prob_flood_risk', 'prob_normal']

for prob in prob_types:
    prob_cols = [f"{city}_{prob}" for city in cities if f"{city}_{prob}" in df.columns]
    if prob_cols:
        df[f'avg_{prob}'] = df[prob_cols].mean(axis=1)
        df[f'max_{prob}'] = df[prob_cols].max(axis=1)
        df[f'min_{prob}'] = df[prob_cols].min(axis=1)
        df[f'std_{prob}'] = df[prob_cols].std(axis=1)

print("Aggregated probability features added.")
df[[f'avg_{p}' for p in prob_types] + [f'max_{p}' for p in prob_types]].head()
```

```python
for class_val, class_name in [(-1, 'drought'), (0, 'normal'), (1, 'flood_risk')]:
    df[f'count_{class_name}'] = (df[actual_cols] == class_val).sum(axis=1)

print("Count features added.")
df[[f'count_{c}' for c in ['drought','normal','flood_risk']]].head()
```

```python
df['usd_x_avg_drought'] = df['USD_LKR_avg'] * df['avg_prob_drought']
df['usd_x_avg_flood']   = df['USD_LKR_avg'] * df['avg_prob_flood_risk']
df['usd_x_avg_normal']  = df['USD_LKR_avg'] * df['avg_prob_normal']

df['week_x_avg_drought'] = df['week_num'] * df['avg_prob_drought']
df['week_x_avg_flood']   = df['week_num'] * df['avg_prob_flood_risk']

print("Interaction features added.")
df[['usd_x_avg_drought', 'usd_x_avg_flood', 'week_x_avg_drought', 'week_x_avg_flood']].head()
```

```python
df['USD_LKR_avg_sq'] = df['USD_LKR_avg'] ** 2
df['RateChange_avg_%_sq'] = df['RateChange_avg'] ** 2
df['week_num_sq'] = df['week_num'] ** 2

print("Polynomial features added.")
df[['USD_LKR_avg_sq', 'RateChange_avg_%_sq', 'week_num_sq']].head()
```

```python
df['week_sin'] = np.sin(2 * np.pi * df['week_num'] / 52)
df['week_cos'] = np.cos(2 * np.pi * df['week_num'] / 52)

print("Cyclical week features added.")
df[['week_num', 'week_sin', 'week_cos']].head(10)
```

```python
veg_dummies = pd.get_dummies(df['vegetable'], prefix='veg', drop_first=True)
df = pd.concat([df, veg_dummies], axis=1)

print("One‑hot encoded vegetable columns added.")
print("New columns:", list(veg_dummies.columns))
df.head()
```

```python
veg_dummy_cols = [col for col in df.columns if col.startswith('veg_')]
for col in veg_dummy_cols:
    df[col] = df[col].astype(int)

print("One‑hot encoded columns after conversion:")
print(df[veg_dummy_cols].head())
print("\nData types:")
print(df[veg_dummy_cols].dtypes)
```

```python
missing = df.isnull().sum()
print("Missing values per column:")
print(missing[missing > 0] if any(missing > 0) else "No missing values found.")
```

```python
# Drop rows with any missing values (mainly from lag and rolling features)
initial_rows = len(df)
df.dropna(inplace=True)
df = df.reset_index(drop=True)
print(f"Dropped {initial_rows - len(df)} rows with missing values.")
print(f"Remaining rows: {len(df)}")
```

```python
missing = df.isnull().sum()
print("Missing values per column:")
print(missing[missing > 0] if any(missing > 0) else "No missing values found.")
```

```python
# Sort by year and week so that all vegetables for a given week appear together
df = df.sort_values(['year', 'week_num']).reset_index(drop=True)

print("Data sorted by year and week.")
df[['year', 'week_num', 'vegetable', 'price']].head(10)

# Drop year column now that sorting is done
df.drop(columns=['year'], inplace=True, errors='ignore')
print("Year column dropped.")
```

```python
output_dir = "../data/feature_engineered"
os.makedirs(output_dir, exist_ok=True)
output_path = os.path.join(output_dir, "vegetable_prices_fe.csv")
df.to_csv(output_path, index=False)
print(f"Enhanced feature set saved to: {output_path}")
```
